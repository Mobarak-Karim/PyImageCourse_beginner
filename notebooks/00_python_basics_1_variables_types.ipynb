# 00 — Python Basics I: Variables, Types, Lists, and Simple Operations

Before image processing, you need to be comfortable reading basic Python. This notebook explains the small set of Python ideas that appear repeatedly in scientific image-analysis code.

## Learning goals
By the end, you should understand:
- what a variable is,
- common Python data types,
- how lists work,
- indexing and slicing,
- arithmetic and comparison operators,
- how to call a function such as `print()` or `len()`.


## 1. Variables: giving a value a useful name
A variable stores a value so we can reuse it later. In scientific code, good names make the analysis easier to understand.


````python
# The equals sign assigns a value to a variable.
pixel_intensity = 145
sample_name = 'cell_01'
is_valid = True

print(pixel_intensity)
print(sample_name)
print(is_valid)
````

### Why this matters
Later you will write variables such as `image`, `threshold`, `mask`, and `labels`. These names are not special Python words; they are names chosen by the programmer.


## 2. Common data types
Python stores different kinds of values using different types. The most common beginner types are:
- `int` — whole numbers,
- `float` — decimal numbers,
- `str` — text,
- `bool` — `True` or `False`.


````python
count = 10
pixel_size_um = 0.65
channel_name = 'green'
use_filter = True

# type(...) tells us the Python type of an object.
print(type(count))
print(type(pixel_size_um))
print(type(channel_name))
print(type(use_filter))
````

## 3. Calling a function
A function performs a task. The general pattern is:

```python
function_name(argument1, argument2)
```

Examples: `print(value)`, `len(values)`, and later `filters.gaussian(image, sigma=2)`.


````python
values = [12, 18, 25, 31]

# len(...) returns the number of items.
number_of_values = len(values)
print(number_of_values)
````

### Positional vs keyword arguments
Later you will often see calls like:

```python
filters.gaussian(image, sigma=2)
```

Here `image` is passed positionally and `sigma=2` is a keyword argument. Keyword arguments make code easier to read because the parameter name is visible.


## 4. Lists
A list stores multiple items in order. Lists are useful for file names, measurements, parameters, or repeated results.


````python
files = ['image_01.tif', 'image_02.tif', 'image_03.tif']

# Python indexing starts at 0.
print(files[0])   # first item
print(files[-1])  # last item

# Slicing selects a range. The stop index is excluded.
print(files[0:2])
````

## 5. Arithmetic and comparisons
Comparisons return Boolean values (`True` or `False`). These become important when we build masks and write logical conditions.


````python
intensity = 180
threshold = 150

print(intensity + 10)
print(intensity > threshold)
print(intensity == threshold)
print(intensity != threshold)
````

## 6. A small scientific example
This combines variables, arithmetic, and function calls.


````python
pixel_size_um = 0.5
object_width_pixels = 24

# Convert a width measured in pixels into micrometers.
object_width_um = object_width_pixels * pixel_size_um

print('Object width:', object_width_um, 'um')
````
