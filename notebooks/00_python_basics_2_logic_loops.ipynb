## 1. Boolean logic
A Boolean expression evaluates to `True` or `False`. In image analysis, this idea becomes the basis of threshold masks.


```python
mean_intensity = 172
minimum_required = 150
maximum_allowed = 220

print(mean_intensity > minimum_required)
print(mean_intensity < maximum_allowed)

# 'and' is True only when BOTH conditions are True.
is_in_range = (mean_intensity > minimum_required) and (mean_intensity < maximum_allowed)
print(is_in_range)
```

### `and`, `or`, `not`
- `A and B` → both must be true.
- `A or B` → at least one must be true.
- `not A` → reverses the Boolean value.

Use parentheses when conditions become longer. They make the logic easier to read.


```python
has_focus = True
is_saturated = False

accept_image = has_focus and (not is_saturated)
print(accept_image)
```

## 2. `if`, `elif`, and `else`
Use a conditional when the program should behave differently depending on a condition.

Python uses **indentation** to define which lines belong to the condition. Four spaces is the standard style.


```python
signal_to_noise = 8.5

if signal_to_noise >= 10:
    print('Strong signal')
elif signal_to_noise >= 5:
    print('Usable signal, but inspect carefully')
else:
    print('Weak signal')
```

### When do we need `if`?
Examples in later scientific code include:
- skip a file if it is not an image,
- choose a processing method based on metadata,
- warn when a mask contains no objects,
- save results only when validation passes.


## 3. `for` loops
A `for` loop repeats a block once for each item in a sequence. This is the main Python pattern for processing multiple image files.


```python
files = ['sample_01.tif', 'sample_02.tif', 'sample_03.tif']

for filename in files:
    # On each iteration, 'filename' refers to one item from the list.
    print('Processing:', filename)
```

### `range()`
`range()` generates integer sequences and is useful when the loop needs an index or a fixed number of repetitions.


```python
for i in range(3):
    print('Iteration:', i)
```

### `enumerate()` is often clearer than managing an index yourself


```python
for index, filename in enumerate(files):
    print(index, filename)
```

## 4. `continue` and `break`
- `continue` skips the rest of the current iteration.
- `break` stops the loop completely.

Use them only when they make the control flow easier to understand.


```python
values = [12, -1, 18, 25]

for value in values:
    if value < 0:
        # Imagine -1 means an invalid measurement. Skip it.
        continue
    print('Valid value:', value)
```

## 5. `while` loops
A `while` loop repeats while a condition remains true. It is useful when the number of iterations is not known in advance.

For ordinary batch image processing, prefer a `for` loop when possible because it is easier to reason about.


```python
attempt = 0

while attempt < 3:
    print('Attempt:', attempt)
    attempt += 1  # same idea as: attempt = attempt + 1
```

## 6. Logic that resembles image thresholding
Later, NumPy lets us apply a condition to every pixel at once. The idea is the same as the simple comparison below.


```python
pixel_values = [80, 120, 180, 210]
threshold = 150

for pixel in pixel_values:
    if pixel > threshold:
        print(pixel, '-> foreground')
    else:
        print(pixel, '-> background')
```
