## 1. Why write a function?
If you repeat the same analysis steps more than once, a function can make the code shorter, clearer, and less error-prone.


````python
def pixels_to_micrometers(length_pixels, pixel_size_um):
    """Convert a length in pixels to micrometers."""
    length_um = length_pixels * pixel_size_um
    return length_um

result = pixels_to_micrometers(24, 0.5)
print(result)
````

### Anatomy of the function
- `def` starts a function definition.
- `pixels_to_micrometers` is the function name.
- `length_pixels` and `pixel_size_um` are parameters.
- the indented block is the function body.
- `return` sends a result back to the caller.


## 2. Parameters vs arguments
A **parameter** is the name in the function definition. An **argument** is the actual value supplied when the function is called.


````python
def describe_intensity(value, threshold=150):
    # threshold has a default value, so the caller may omit it.
    if value > threshold:
        return 'bright'
    return 'dark'

print(describe_intensity(180))
print(describe_intensity(180, threshold=200))
````

### Why keyword arguments are useful
`threshold=200` is easier to understand than a bare second number because the meaning of the argument is visible. This is why scientific Python code often uses keyword arguments heavily.


## 3. Local variables
Variables created inside a function normally belong to that function. This helps prevent accidental interference between different parts of a program.


````python
def area_from_pixels(area_pixels, pixel_size_um):
    pixel_area_um2 = pixel_size_um ** 2
    area_um2 = area_pixels * pixel_area_um2
    return area_um2

print(area_from_pixels(100, 0.5))
````

## 4. Imports and modules
Python libraries are organized into modules. We import the functionality we need instead of rewriting it ourselves.


````python
import math

# Dot notation means: use sqrt from the math module.
print(math.sqrt(25))
````

### Common import styles you will see later


Then we call functions using names such as `np.mean(...)`, `plt.imshow(...)`, or `filters.gaussian(...)`.


## 5. Reading an error message
Errors are normal. The important skill is learning to read the final line of the traceback first.

Common examples:
- `NameError` — a variable/function name does not exist.
- `TypeError` — an operation received the wrong kind of object.
- `IndexError` — an index is outside the available range.
- `FileNotFoundError` — Python cannot find the requested file.
- `ModuleNotFoundError` — a package is missing or the wrong environment is active.


````python
values = [10, 20, 30]

# Correct indices are 0, 1, and 2.
print(values[2])

# If you intentionally try values[3], Python raises IndexError.
# Keeping the incorrect line commented lets this teaching notebook run cleanly.
````

## 6. `try/except`: use only when you can handle the failure meaningfully
Do not hide every error with `try/except`. Use it when there is a known situation that your program can respond to safely.


````python
text_value = '12.5'

try:
    number = float(text_value)
    print(number)
except ValueError:
    print('The text could not be converted to a number.')
````

## 7. A reusable pattern for scientific code
A useful function should usually have:
1. a clear name,
2. explicit inputs,
3. a focused job,
4. a return value when another step needs the result,
5. parameters that can be changed without editing the function body.


````python
def classify_measurement(value, low=10, high=20):
    """Classify one measurement using explicit limits."""
    if value < low:
        return 'low'
    elif value > high:
        return 'high'
    else:
        return 'within range'

print(classify_measurement(16))
````
