## 1. Histograms describe intensity distribution

A histogram helps answer whether intensities occupy a narrow range, whether outliers are present, and whether a simple threshold might be plausible.

```python
import numpy as np
import matplotlib.pyplot as plt
import skimage as ski
from skimage import exposure
image = ski.data.moon()
# ravel() flattens all pixels into one vector for the histogram.
plt.hist(image.ravel(), bins=256)
plt.xlabel("Intensity"); plt.ylabel("Pixel count"); plt.show()
```

## 2. Contrast stretching

Percentiles are useful intensity landmarks because extreme outliers affect them less than absolute minima and maxima.

```python
# Use 2nd and 98th percentiles as robust display/processing landmarks.
p2, p98 = np.percentile(image, (2, 98))
# This creates a NEW remapped array; unlike vmin/vmax it changes pixel values.
stretched = exposure.rescale_intensity(image, in_range=(p2, p98))
fig, ax = plt.subplots(1,2,figsize=(10,4))
ax[0].imshow(image,cmap="gray"); ax[0].set_title("Original")
ax[1].imshow(stretched,cmap="gray"); ax[1].set_title("2–98% stretch")
for a in ax: a.axis("off")
plt.show()
```

## 3. Global versus local enhancement

Use local enhancement only when spatially varying contrast is the actual problem; it can also amplify noise.

```python
# Global equalization uses one mapping for the whole image.
global_eq = exposure.equalize_hist(image)
# CLAHE adapts locally; clip_limit limits local amplification.
local_eq = exposure.equalize_adapthist(image, clip_limit=0.03)
fig, ax = plt.subplots(1,3,figsize=(12,4))
for a,im,t in zip(ax,[image,global_eq,local_eq],["Original","Global equalization","CLAHE"]):
    a.imshow(im,cmap="gray"); a.set_title(t); a.axis("off")
plt.show()
```

## Function-selection guide

| Problem | Function | Caution |
|---|---|---|
| Inspect intensity distribution | `plt.hist()` | Histogram is descriptive, not segmentation |
| Display only | `vmin`, `vmax` | Array remains unchanged |
| Explicit remapping | `rescale_intensity()` | Changes pixel values |
| Local contrast variation | `equalize_adapthist()` | May amplify noise |

## Takeaway

**Choose functions because they solve a specific image problem, and always inspect the result before measuring.**
