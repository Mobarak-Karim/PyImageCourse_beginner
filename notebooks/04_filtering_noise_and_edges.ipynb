## 1. Filtering should solve a specific problem

Do not smooth automatically. Filtering can remove noise, but it can also erase small real structures or merge neighboring objects.

```python
import matplotlib.pyplot as plt
import skimage as ski
from skimage import filters, util, morphology
image = ski.data.camera()
# Add controlled noise only for teaching comparison.
gaussian_noise = util.random_noise(image, mode="gaussian", var=0.005)
salt_pepper = util.random_noise(image, mode="s&p", amount=0.03)
```

## 2. Gaussian versus median

Gaussian smoothing is a good first choice for fine distributed noise. Median filtering is especially useful for isolated bright/dark outliers.

```python
# sigma controls Gaussian blur scale: larger sigma removes more detail.
gaussian_filtered = filters.gaussian(gaussian_noise, sigma=1.2)
# The footprint controls the median neighborhood.
median_filtered = filters.median(salt_pepper, footprint=morphology.disk(1))
fig, ax=plt.subplots(2,2,figsize=(10,8))
for a,im,t in zip(ax.ravel(),[gaussian_noise,gaussian_filtered,salt_pepper,median_filtered],["Gaussian noise","Gaussian filter","Salt & pepper","Median filter"]):
    a.imshow(im,cmap="gray"); a.set_title(t); a.axis("off")
plt.show()
```

## 3. Edge detection

Sobel estimates local intensity change. An edge image is useful for boundaries, but it is not the same as a filled segmentation mask.

```python
# Sobel highlights rapid intensity transitions.
edges = filters.sobel(image)
fig, ax=plt.subplots(1,2,figsize=(10,4))
ax[0].imshow(image,cmap="gray"); ax[0].set_title("Image")
ax[1].imshow(edges,cmap="magma"); ax[1].set_title("Sobel edge strength")
for a in ax: a.axis("off")
plt.show()
```

## Function-selection guide

| Problem | Function | Main parameter |
|---|---|---|
| Fine distributed noise | `filters.gaussian()` | `sigma` |
| Isolated impulse noise | `filters.median()` | `footprint` |
| Boundary/gradient strength | `filters.sobel()` | none |

**Check afterward:** did noise decrease without erasing or joining real structures?

## Takeaway

**Choose functions because they solve a specific image problem, and always inspect the result before measuring.**
