## 1. A mask is not yet an object table

`measure.label()` assigns an integer ID to each connected foreground region. Background is usually label 0.

```python
import matplotlib.pyplot as plt
import skimage as ski
from skimage import filters, measure, color, morphology
image = ski.data.coins()
smooth = filters.gaussian(image, sigma=1)
mask = smooth > filters.threshold_otsu(smooth)
mask = morphology.remove_small_objects(mask, max_size=80)
# Each connected foreground component receives a unique integer label.
labels = measure.label(mask)
print("labeled regions:", labels.max())
```

## 2. Validate with an overlay

Colored labels over the source image reveal missed objects, false positives, merges, and splits.

```python
# label2rgb creates a visual QC overlay; alpha controls transparency.
overlay = color.label2rgb(labels, image=image, bg_label=0, alpha=0.35)
fig, ax=plt.subplots(1,2,figsize=(10,4))
ax[0].imshow(mask,cmap="gray"); ax[0].set_title("Binary mask")
ax[1].imshow(overlay); ax[1].set_title("Label overlay")
for a in ax: a.axis("off")
plt.show()
```

## Function-selection guide

| Need | Function | Check |
|---|---|---|
| Connected object IDs | `measure.label()` | Connectivity and mask correctness |
| Visual QC | `color.label2rgb()` | Do labels match the source? |
| Split touching objects | `segmentation.watershed()` | Are markers meaningful? |

## Takeaway

**Choose functions because they solve a specific image problem, and always inspect the result before measuring.**
