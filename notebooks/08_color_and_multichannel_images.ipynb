## 1. Channel meaning comes from acquisition context

An array shaped `(rows, cols, 3)` might be RGB, or three independent fluorescence channels. Do not infer biology from axis length alone.

```python
import numpy as np
import matplotlib.pyplot as plt
import skimage as ski
from skimage import color, filters, measure
rgb = ski.data.astronaut()
# Extract RGB channels by final-axis index.
r,g,b = rgb[...,0], rgb[...,1], rgb[...,2]
fig, ax=plt.subplots(1,3,figsize=(12,4))
for a,im,t in zip(ax,[r,g,b],["Red","Green","Blue"]): a.imshow(im,cmap="gray"); a.set_title(t); a.axis("off")
plt.show()
```

## 2. RGB conversion versus channel extraction

Use `rgb2gray()` for true RGB color data. For scientific multichannel data, usually analyze the relevant channel directly.

```python
# rgb2gray applies a color-luminance conversion for true RGB data.
gray = color.rgb2gray(rgb)
plt.imshow(gray,cmap="gray"); plt.title("RGB converted to grayscale"); plt.axis("off"); plt.show()
```

## 3. Synthetic two-channel scientific example

Synthetic data keeps the lesson reproducible and avoids exposing research data.

```python
# Build two independent synthetic fluorescence-style channels.
y,x = np.mgrid[:256,:256]
ch1 = np.exp(-((x-90)**2+(y-120)**2)/(2*25**2))
ch2 = np.exp(-((x-165)**2+(y-135)**2)/(2*35**2))
stack = np.stack([ch1,ch2], axis=-1)
print("multichannel shape:", stack.shape)
# Analyze channel 0 because the analytical question defines it as the target channel.
t = filters.threshold_otsu(stack[...,0])
labels = measure.label(stack[...,0] > t)
print("channel-0 objects:", labels.max())
```

## Function-selection guide

| Data type | Action |
|---|---|
| True RGB photograph | `color.rgb2gray()` when grayscale is analytically appropriate |
| Independent fluorescence channels | Extract specific channel: `image[..., i]` |
| Unknown channel semantics | Check metadata/acquisition documentation first |

## Takeaway

**Choose functions because they solve a specific image problem, and always inspect the result before measuring.**
